```python
from datasets import load_dataset 
```

```python
split = "validation"

ds = load_dataset(
    "imagenet-1k",
    split=split,
    cache_dir=None,
    trust_remote_code=True,
)
total = len(ds)
print(f"  {total:,} examples")
```

```python
for item in ds.select(range(5)):  # returns list of dicts
    label = item["label"]
    img = item["image"]
    print(label)
    print(img.size)
```
